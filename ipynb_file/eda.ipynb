```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df= pd.read_excel("../raw_data/Bank_Cust_Sub.xlsx")
df.head()
```

```python
df.drop(columns="Customer_ID", inplace=True)
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
numeric_cols = [
    "Age",
    "Annual_Income",
    "Credit_Score",
    "Account_Balance",
    "Campaign_Count",
    "Months_With_Bank"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
    df[col] = df[col].fillna(df[col].median())
    df[col]=df[col].astype(int)
```

```python
col_cat=["Job", "Education", "Housing_Loan", "Personal_Loan", "Contacted_Before" ]

for col in col_cat:
    df[col]=df[col].str.strip().str.title()
    df[col]=df[col].fillna(df[col].mode()[0])
df.isna().sum()
```
